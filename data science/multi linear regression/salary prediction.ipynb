```python
import pandas as pd
import matplotlib.pyplot as plt
from sklearn.linear_model import LinearRegression
from word2number import w2n
```

```python
data = pd.read_csv(r"hiring.csv")
data
```

```python
data.experience = data.experience.fillna('zero')
data
```

```python
data.experience = data.experience.apply(w2n.word_to_num)
data
```

```python
data = data.rename(columns={'test_score(out of 10)':'test_score', 'interview_score(out of 10)':'interview_score'})
data
```

```python
# fill test_score , null value with avg score
data.test_score.median()
```

```python
data.test_score.mean()
```

```python
import math
```

```python
avg_score = math.floor(data.test_score.mean())
avg_score
```

```python
data.test_score = data.test_score.fillna(avg_score)
data
```

```python
X = data.drop('salary($)', axis='columns')
X
```

```python
y = data['salary($)']
y
```

```python
lr = LinearRegression()
lr.fit(X,y)
```

```python
lr.predict([[2,9,6]])
```
